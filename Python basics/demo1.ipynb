Write code yourself. Given scores = [55, 90, 72, 38, 81], use a loop to compute the sum of all scores, and a second accumulator for how many scores are below 60. Print both. Name the loop variable so it matches what it holds, and predict both numbers before running.

```python
scores = [55, 90, 72, 38, 81]
total = 0
for i in scores:
    total = total + i
    print(total)
count = 0
for j in scores:
    if j < 60:
        count += 1
print("Count:",count)
```
